```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
df['Unit_Price'].fillna(df['Unit_Price'].median(), inplace=True)
df['City'].fillna("Unknown", inplace=True)
df['Quantity'].fillna(df['Quantity'].median(), inplace=True)

```

```python
df.isnull().sum()
```

```python
print(df[df.duplicated(keep=False)])
```

```python
df.drop_duplicates(inplace=True)
```

```python
numerical_cols = df.select_dtypes(include=['number']).columns
for col in numerical_cols:
    try:
        df[col] = pd.to_numeric(df[col], errors='ignore')
    except:
        pass
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = df[col].str.strip().str.lower()

df.to_csv('data.csv', index=False)
```

```python
df.head()
```

```python
cols=['Age', 'Unit_Price', 'Total_Amount']

for col in cols:
    mean=df[col].mean()
    median=df[col].median()
    diff = abs(mean-median)

    mode=df[col].mode()[0]

    print(f"Mean of {col}:{mean} Median of {col}:{median} Mode of {col}:{mode}")
    
    if diff > 0:
        print(f"{col} is skewed.")
    elif diff< 0:
        print(f"{col} is not skewed.")
    else:
        print(f"{col} is perfectly symmetric.")

    print("\n")
```

```python
cols=['Age', 'Unit_Price', 'Total_Amount']

for col in cols:
    q1= df[col].quantile(0.25)
    q3= df[col].quantile(0.75)  

    irq=q3-q1

    lower_bound=q1-1.5*irq
    upper_bound=q3+1.5*irq

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    print(f"Outliers in {col}:\n{outliers}\n")

    # df=df[df[col>=lower_bound] & df[col]<=upper_bound ]

    df[col] = df[col].clip(lower_bound, upper_bound)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python

df[["Unit_Price","Quantity","Total_Amount"]] = MinMaxScaler().fit_transform(df[["Unit_Price","Quantity","Total_Amount"]])
df['Age'] = StandardScaler().fit_transform(df[['Age']])
```

```python
df = pd.get_dummies(df, columns=['Gender', 'Product_Category', 'City'], drop_first=True)
```
